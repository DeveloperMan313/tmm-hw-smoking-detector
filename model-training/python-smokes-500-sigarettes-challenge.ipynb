```python
from mediapipe_model_maker import object_detector
```

```python
train_data = object_detector.Dataset.from_pascal_voc_folder('voc_cigarette_mouth/train')
val_data = object_detector.Dataset.from_pascal_voc_folder('voc_cigarette_mouth/test')
```

```python
spec = object_detector.SupportedModels.MOBILENET_MULTI_AVG
```

```python
hparams = object_detector.HParams(
    export_dir='./cigarette_model',
    epochs=50
)
```

```python
options = object_detector.ObjectDetectorOptions(
    supported_model=spec,
    hparams=hparams
)
```

```python
model = object_detector.ObjectDetector.create(
    train_data=train_data,
    validation_data=val_data,
    options=options
)
```

```python
eval_result = model.evaluate(val_data)
print(f"mAP: {eval_result}")
```

```python
model.export_model()
```
